```python
#Exercise 1 Create the following list of Strings: [“Patryk”, “Alex”, “Ian”, “Kevin”]
list_of_strings = ["Patryk", "Alex", "Ian", "Kevin"]
```

```python
#A Prints out all values separately
for element in list_of_strings:
  print(element)
```

```python
#B Prints out second value
print (list_of_strings[1]) #alternatively just: list_of_strings[1]
```

```python
#C Checks the length of a list
len(list_of_strings)
```

```python
#D Replaces all names with a word “Tomato”
for element in list_of_strings:
  print("Tomato")
```

```python
#Exercise 2 Create the following list of Integers: [2, 4, 8, 12, 1, 67, 31]
list_of_integers = [2, 4, 8, 12, 1, 67, 31]
```

```python
#A Converts all two-digit numbers into any object of String type

list_with_strings = [str(num) if 10 <= num <= 99 else num for num in list_of_integers]
print(list_with_strings)
```

```python
#B Prints the lowest value

lowest_value = min(list_of_integers)
print(lowest_value)
```
